```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

df = pd.read_csv("experiment.csv")
df = df.loc[df.epoch != 0]

# Create an interactive plot
fig = px.line(
    df,
    x="epoch",
    y="separation",
    color="scramble_distance",
    title="Interactive Plot: Epoch vs. Separation",
    labels={"scramble_distance": "Scramble Distance"},
)
fig.show()
```
